## 1. Espaço de Busca — quantas soluções existem?

Antes de resolver um problema com IA, precisamos entender **quão grande é o espaço de busca**.

```python
# Exemplo: problema das N-rainhas
# Quantas configurações possíveis existem para N rainhas em um tabuleiro NxN?

def espaco_busca_rainhas(n):
    """Calcula o tamanho do espaço de busca bruto (permutações de colunas)."""
    from math import factorial
    return factorial(n)

for n in range(1, 13):
    tamanho = espaco_busca_rainhas(n)
    print(f"N={n:2d} → {tamanho:>20,} configurações")
```

## 2. Paisagem de Aptidão (*Fitness Landscape*)

A paisagem de aptidão representa o valor de cada solução candidata. Um **ótimo global** é o pico mais alto; **ótimos locais** são picos menores.

```python
import numpy as np
import matplotlib.pyplot as plt

# Função com múltiplos ótimos locais
def paisagem(x):
    return np.sin(3*x) * np.exp(-0.1*x**2) + 0.5 * np.cos(x)

x = np.linspace(-5, 5, 500)
y = paisagem(x)

plt.figure(figsize=(10, 4))
plt.plot(x, y, 'b-', linewidth=2, label='Paisagem de Aptidão')
idx_global = np.argmax(y)
plt.scatter([x[idx_global]], [y[idx_global]], color='red', s=120, zorder=5, label=f'Ótimo global ≈ {x[idx_global]:.2f}')
plt.xlabel('Parâmetro (solução candidata)')
plt.ylabel('Aptidão')
plt.title('Exemplo de Paisagem de Aptidão com Ótimos Locais')
plt.legend(); plt.grid(True); plt.show()
print(f"Valor do ótimo global: x = {x[idx_global]:.4f}, f(x) = {y[idx_global]:.4f}")
```

## 3. Complexidade de Problemas — Big-O

Vamos comparar empiricamente o tempo de execução de algoritmos O(n), O(n²) e O(n³).

```python
import time

def algoritmo_linear(n):
    s = 0
    for i in range(n): s += i
    return s

def algoritmo_quadratico(n):
    s = 0
    for i in range(n):
        for j in range(n): s += i+j
    return s

tamanhos = [100, 500, 1000, 2000, 4000]
tempos_lin, tempos_quad = [], []

for n in tamanhos:
    t0 = time.perf_counter(); algoritmo_linear(n);    tempos_lin.append(time.perf_counter()-t0)
    t0 = time.perf_counter(); algoritmo_quadratico(n); tempos_quad.append(time.perf_counter()-t0)

plt.figure(figsize=(9,4))
plt.plot(tamanhos, tempos_lin,  'bo-', label='O(n) — linear')
plt.plot(tamanhos, tempos_quad, 'rs-', label='O(n²) — quadrático')
plt.xlabel('Tamanho da entrada (n)'); plt.ylabel('Tempo (s)')
plt.title('Comparação de Complexidade'); plt.legend(); plt.grid(True); plt.show()
```

## 4. Vizinhança de Soluções

Em otimização, a **vizinhança** de uma solução são as soluções alcançáveis em um passo de modificação.

```python
# Exemplo: vizinhança de uma sequência binária (flip de 1 bit)
def vizinhanca_binaria(solucao):
    """Retorna todos os vizinhos com 1 bit diferente."""
    vizinhos = []
    for i in range(len(solucao)):
        vizinho = list(solucao)
        vizinho[i] = 1 - vizinho[i]  # flip do bit i
        vizinhos.append(tuple(vizinho))
    return vizinhos

solucao = (0, 1, 0, 1)
vizinhos = vizinhanca_binaria(solucao)
print(f"Solução atual:  {solucao}")
print(f"Número de vizinhos: {len(vizinhos)}")
for i, v in enumerate(vizinhos):
    print(f"  Vizinho {i+1}: {v}  ← bit {i} invertido")
```
