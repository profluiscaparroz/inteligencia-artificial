## 1. Análise de Crescimento da IA — Dados e Visualizações

## 2. Ética em IA — Análise de Viés

```python
import numpy as np

# Simulação de um dataset com viés histórico
np.random.seed(42)

# Dataset sintético: contratação de funcionários
# Features: nota_tecnica (0-10), anos_exp (0-10)
# Label: contratado (1) ou não (0)
# Viés: historicamente, homens foram mais contratados

n = 200
genero = np.random.choice([0,1], n)  # 0=mulher, 1=homem
nota   = np.random.uniform(5, 10, n)
exp    = np.random.uniform(0, 10, n)

# Decisão com viés: homens têm 20% mais chance
prob_base = (nota/10)*0.5 + (exp/10)*0.3
prob_viesada = np.clip(prob_base + genero*0.2, 0, 1)
contratado = (np.random.random(n) < prob_viesada).astype(int)

# Análise do viés
taxa_homem  = contratado[genero==1].mean()
taxa_mulher = contratado[genero==0].mean()

print("=== Análise de Viés no Dataset ===")
print(f"Taxa de contratação — Homens:  {taxa_homem:.1%}")
print(f"Taxa de contratação — Mulheres: {taxa_mulher:.1%}")
print(f"Diferença:                      {(taxa_homem-taxa_mulher)*100:.1f} p.p.")
print()

# Distribuições de notas por gênero
print("Nota técnica média — Homens:  ", nota[genero==1].mean().round(2))
print("Nota técnica média — Mulheres:", nota[genero==0].mean().round(2))
print("Anos de exp. médio — Homens:  ", exp[genero==1].mean().round(2))
print("Anos de exp. médio — Mulheres:", exp[genero==0].mean().round(2))
print()
print("⚠️  Apesar de qualificações similares, homens são contratados mais.")
print("   Um modelo ML treinado nesse dataset vai perpetuar o viés!")

# Visualização
fig, axes = plt.subplots(1,2,figsize=(12,5))
for g,rotulo,cor in [(0,'Mulheres','red'),(1,'Homens','blue')]:
    mask = genero==g
    axes[0].scatter(nota[mask],exp[mask],c=cor,label=rotulo,
                   marker='o' if contratado[mask].mean()>0.5 else 'x',alpha=0.5)
axes[0].set_xlabel('Nota Técnica'); axes[0].set_ylabel('Anos de Exp.')
axes[0].set_title('Dataset de Contratação (com viés)'); axes[0].legend()
axes[1].bar(['Mulheres','Homens'],[taxa_mulher,taxa_homem],color=['red','blue'],alpha=0.7)
axes[1].set_ylabel('Taxa de Contratação'); axes[1].set_title('Disparidade na Contratação'); axes[1].set_ylim(0,1)
for i,(t,lab) in enumerate([(taxa_mulher,'Mulheres'),(taxa_homem,'Homens')]):
    axes[1].text(i, t+0.02, f'{t:.0%}', ha='center', fontsize=12)
plt.tight_layout(); plt.show()
```

## 3. Regulação e Impacto Social — Mapa Mental

```python
# Visualização: dimensões de impacto da IA
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

dimensoes = {
    "Saúde": {"impacto":9,"risco":7,"maturidade":6,"cor":"green"},
    "Educação": {"impacto":8,"risco":5,"maturidade":5,"cor":"blue"},
    "Trabalho": {"impacto":9,"risco":8,"maturidade":7,"cor":"orange"},
    "Segurança": {"impacto":7,"risco":9,"maturidade":6,"cor":"red"},
    "Finanças": {"impacto":8,"risco":7,"maturidade":8,"cor":"purple"},
    "Transporte": {"impacto":7,"risco":6,"maturidade":7,"cor":"brown"},
    "Meio Ambiente": {"impacto":8,"risco":4,"maturidade":4,"cor":"teal"},
    "Entretenimento": {"impacto":6,"risco":3,"maturidade":8,"cor":"pink"},
}

fig, ax = plt.subplots(figsize=(10,8))
for setor, dados in dimensoes.items():
    ax.scatter(dados["risco"], dados["impacto"],
               s=dados["maturidade"]*80,
               c=dados["cor"], alpha=0.7, edgecolors='black', linewidths=1.5)
    ax.annotate(setor, (dados["risco"]+0.1, dados["impacto"]+0.1), fontsize=10)

ax.set_xlabel('Risco / Preocupação Ética (1-10)', fontsize=12)
ax.set_ylabel('Impacto Potencial da IA (1-10)', fontsize=12)
ax.set_title('Mapa de Impacto da IA por Setor\n(tamanho = maturidade tecnológica)', fontsize=13)
ax.set_xlim(0,11); ax.set_ylim(0,11)
ax.axhline(7,color='gray',linestyle='--',alpha=0.4)
ax.axvline(7,color='gray',linestyle='--',alpha=0.4)
ax.text(7.5,9,'Alto Impacto\nAlto Risco',fontsize=9,color='red',alpha=0.7)
ax.text(1,9,'Alto Impacto\nBaixo Risco',fontsize=9,color='green',alpha=0.7)
ax.grid(True,alpha=0.3); plt.tight_layout(); plt.show()
```
