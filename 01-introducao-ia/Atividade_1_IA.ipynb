# Exercício prático (Criando uma IA simples)

````python
especie = "Humano"

dados = [
    (2, "Infantil"),
    (3, "Infantil"),
    (4, "Infantil"),
    (10, "Adulto"),
    (20, "Adulto"),
    (30, "Adulto"),
    (40, "Adulto"),
    (52, "Idoso"),
    (60, "Idoso"),
    (70, "Idoso"),
    (80, "Idoso"),
]

# -------------------------------
# Organizar dados por classe
# -------------------------------

classes = {}

for idade, classe in dados:
    if classe not in classes:
        classes[classe] = []
    classes[classe].append(idade)

# -------------------------------
# Calcular médias
# -------------------------------

medias = {}

for classe in classes:
    soma = 0
    for idade in classes[classe]:
        soma += idade

    medias[classe] = soma / len(classes[classe])

# -------------------------------
# Ordenar classes pela média
# -------------------------------

medias_ordenadas = sorted(medias.items(), key=lambda x: x[1])

# -------------------------------
# Calcular limites entre classes
# -------------------------------

limites = []

for i in range(len(medias_ordenadas) - 1):

    classe_atual, media_atual = medias_ordenadas[i]
    classe_proxima, media_proxima = medias_ordenadas[i + 1]

    limite = (media_atual + media_proxima) / 2

    limites.append((limite, classe_proxima))

# -------------------------------
# Função de previsão
# -------------------------------

def prever(idade):

    classe_prevista = medias_ordenadas[0][0]

    for limite, proxima_classe in limites:

        if idade >= limite:
            classe_prevista = proxima_classe

    return classe_prevista

# -------------------------------
# Mostrar o que a IA "aprendeu"
# -------------------------------

print("Especie:", especie)

print("\nMedias aprendidas:")
for classe, media in medias_ordenadas:
    print(classe, "->", media)

print("\nLimites aprendidos:")
for i, (limite, classe) in enumerate(limites, start=1):
    print("Limite", i, "=", limite, "| a partir daqui tende para:", classe)

# -------------------------------
# Testar previsões
# -------------------------------

idade_teste = [1, 5, 10, 20, 40, 60, 80, 100]

print("\nPrevisões:")
for idade in idade_teste:
    print("Idade", idade, "->", prever(idade))
````

````python
import pandas as pd

especie = "humano"

# -------------------------------
# Carregar dados usando Pandas
# -------------------------------
try:
    df = pd.read_csv('/content/dados.csv', delimiter=";")
    # Garante que as colunas existam e converte para a lista de tuplas esperada pelo restante do código
    dados = list(df[['idade', 'classe']].itertuples(index=False, name=None))
except FileNotFoundError:
    print("Erro: O arquivo 'dados.csv' não foi encontrado. Usando dados de exemplo.")
    dados = [(2, "Infantil"), (10, "Adulto"), (60, "Idoso")]
except Exception as e:
    print(f"Erro ao ler o arquivo: {e}")
    dados = []

# -------------------------------
# Organizar dados por classe
# -------------------------------

classes = {}

for idade, classe in dados:
    if classe not in classes:
        classes[classe] = []
    classes[classe].append(idade)

# -------------------------------
# Calcular médias
# -------------------------------

medias = {}

for classe in classes:
    soma = sum(classes[classe])
    medias[classe] = soma / len(classes[classe])

# -------------------------------
# Ordenar classes pela média
# -------------------------------

medias_ordenadas = sorted(medias.items(), key=lambda x: x[1])

# -------------------------------
# Calcular limites entre classes
# -------------------------------

limites = []

for i in range(len(medias_ordenadas) - 1):
    classe_atual, media_atual = medias_ordenadas[i]
    classe_proxima, media_proxima = medias_ordenadas[i + 1]
    limite = (media_atual + media_proxima) / 2
    limites.append((limite, classe_proxima))

# -------------------------------
# Função de previsão
# -------------------------------

def prever(idade):
    if not medias_ordenadas:
        return "Sem dados"

    classe_prevista = medias_ordenadas[0][0]
    for limite, proxima_classe in limites:
        if idade >= limite:
            classe_prevista = proxima_classe
    return classe_prevista

# -------------------------------
# Mostrar o que a IA "aprendeu"
# -------------------------------

print("Especie:", especie)

if medias_ordenadas:
    print("\nMedias aprendidas:")
    for classe, media in medias_ordenadas:
        print(f"{classe} -> {media:.2f}")

    print("\nLimites aprendidos:")
    for i, (limite, classe) in enumerate(limites, start=1):
        print(f"Limite {i} = {limite} | a partir daqui tende para: {classe}")

    # -------------------------------
    # Testar previsões
    # -------------------------------
    idade_teste = [1, 5, 10, 20, 40, 60, 80, 100]
    print("\nPrevisões:")
    for idade in idade_teste:
        print(f"Idade {idade} -> {prever(idade)}")
````
