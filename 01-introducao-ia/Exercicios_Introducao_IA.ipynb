Existem quatro grandes paradigmas: **Simbólico**, **Conexionista**, **Evolucionário** e **Probabilístico**.

## 3. Agente Inteligente Simples — Agente Reativo

```python
# Um agente reativo simula um robô aspirador em um grid 1D
import random

class AgenteAspirador:
    """Agente reativo simples que limpa sujeira em um corredor."""
    
    def __init__(self, n_salas=5, sujeira_inicial=0.6):
        self.salas = ['S' if random.random() < sujeira_inicial else 'L'
                      for _ in range(n_salas)]
        self.posicao = 0
        self.limpezas = 0
        self.passos = 0
    
    def perceber(self):
        return self.salas[self.posicao]
    
    def agir(self):
        percepcao = self.perceber()
        if percepcao == 'S':
            self.salas[self.posicao] = 'L'
            self.limpezas += 1
            return "LIMPAR"
        elif self.posicao < len(self.salas) - 1:
            self.posicao += 1
            self.passos += 1
            return "MOVER →"
        else:
            self.posicao = 0
            self.passos += 1
            return "VOLTAR ao início"
    
    def executar(self, max_ciclos=20):
        print(f"Estado inicial: {self.salas}  (S=sujo, L=limpo)")
        for ciclo in range(max_ciclos):
            acao = self.agir()
            print(f"Ciclo {ciclo+1:2d} | Pos: {self.posicao} | Ação: {acao:15s} | {self.salas}")
            if all(s == 'L' for s in self.salas):
                print(f"\n✅ Tudo limpo em {ciclo+1} ciclos ({self.limpezas} limpezas, {self.passos} movimentos)")
                return
        print("⚠️  Limite de ciclos atingido.")

random.seed(42)
agente = AgenteAspirador(n_salas=6)
agente.executar()
```
